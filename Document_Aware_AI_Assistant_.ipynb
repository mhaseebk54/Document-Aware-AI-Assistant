```python
import pdfplumber

def load_document(file_path):
    text = ""
    if file_path.endswith(".txt"):
        with open(file_path, "r", encoding="utf-8") as f:
            text = f.read()
    elif file_path.endswith(".pdf"):
        with pdfplumber.open(file_path) as pdf:
            for page in pdf.pages:
                text += page.extract_text() + " "
    else:
        raise ValueError("Unsupported file type. Use .txt or .pdf")


    text = ' '.join(text.split())
    return text

document_text = load_document("/content/document.txt")
print(document_text[:500], "...")
```

```python
from transformers import pipeline

model_name = "google/flan-t5-base"
qa_pipeline = pipeline("text2text-generation", model=model_name, device=-1)
```

```python
def generate_prompt(document, question):
    prompt = f"""
You are a helpful assistant. Use only the following document to answer the question.
If the answer is not in the document, reply with "Answer not found in the document."

Document:
{document}

Question:
{question}

Answer:
"""
    return prompt
```

```python
def answer_question(document, question):
    prompt = generate_prompt(document, question)
    result = qa_pipeline(prompt, max_length=200)[0]['generated_text']
    return result.strip()


user_question = "What is the main topic of the document?"
answer = answer_question(document_text, user_question)

print("Question:", user_question)
print("Answer:", answer)
print("Model Used:", model_name)
```
